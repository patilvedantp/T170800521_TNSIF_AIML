````python
#kwargs functions : Keyboard argument function
def display(**details):  # **kwargs :allows a function to accept any number of keyword arguments. All arguments are collected into a dictionary, where the argument names become keys and their corresponding values become dictionary values.
  print(details)

display(name="Vedant",age=21,address="Pune")
display(name="Ganesh",age="Ganesh",address="Mumbai")

display()
````

````python
class Dog:
  species = "Canine" #Class Attribute

  def __init__(self, name, age):
    self.name = name #Instance Attribute
    self.age = age  #Instance Attribute

#Creating an object of a Dog Class
dog1 = Dog("Buddy",5)
dog2 = Dog("Max",6)

print(dog1.name)
print(dog2.age)
print(dog1.species)
print(dog2.species)
````

````python
class Dog:
  species = "Canine"  # Class attribute

  def __init__(self,name,age):
    self.name = name
    self.age = age

  def animal_name(self, dog, cat):
    self.dog = dog
    self.cat = cat
    print("Dog Name:", self.dog)
    print("Cat Name:", self.cat)
    print("Dog name from init function: ",self.name)

  def location(self, state, country):
    self.state = state
    self.country = country
    print("State name: ", self.state)
    print("Country Name: ", self.country)

# Taking input from user
name = input("Enter Dog Name (Object): ")
age = int(input("Enter Dog age: "))

dog_value = input("Enter Dog Name: ")
cat_value = input("Enter Cat Value: ")

# Creating object
d = Dog(name, age)
print(d.name)
print(d.age)
print(d.species)

# Calling function with user input
d.animal_name(dog_value, cat_value)

d.location("Maharashtra", "India")
````

````python
class car:
  def __init__(self, brand):
    self.brand = brand

  def show(self, colour):
    self.colour = colour
    print("Brand of the car is: ", self.brand)
    print("Colour of the car is: ", self.colour)

  def display(self):
    print("Brand: ", self.brand)
    print("Colour: ", self.colour)

c1 = car("BMW")
c1.show("Red")
c1.display()
````
