1. Check even or odd

```python
def check_even_odd(number):
    if number % 2 == 0:
        return 'Even'
    else:
        return 'Odd'

# Test cases
numbers = [10, 15, 0, -3]
for num in numbers:
    print(f"{num} is {check_even_odd(num)}")
```

### 5. Find the factorial of a number using a loop

```python
def factorial_loop(n):
    if n < 0:
        return None
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print(f"Factorial of 6: {factorial_loop(6)}")
```

### 7. Reverse a number using a loop

```python
def reverse_number(n):
    sign = -1 if n < 0 else 1
    n = abs(n)
    reversed_num = 0
    while n > 0:
        reversed_num = (reversed_num * 10) + (n % 10)
        n //= 10
    return sign * reversed_num

print(f"Reversed 12345: {reverse_number(12345)}")
```

### 8. Check whether a number is prime

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print(f"Is 17 prime?: {is_prime(17)}")
```

### 9. Write a function to check if a number is palindrome

```python
def is_palindrome(n):
    return str(n) == str(n)[::-1]

print(f"Is 121 a palindrome?: {is_palindrome(121)}")
```

### 10. Write a function to calculate sum of digits of a number

### 14. Reverse a list without using built-in functions

```python
def reverse_list_custom(lst):
    left, right = 0, len(lst) - 1
    while left < right:
        lst[left], lst[right] = lst[right], lst[left]
        left += 1
        right -= 1
    return lst

print(f"Reversed list: {reverse_list_custom([1, 2, 3, 4])}")
```

### 23. Count frequency of characters in a string

```python
def char_frequency(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    return freq

print(f"Char frequency in 'hello': {char_frequency('hello')}")
```

### 27. Create a class Rectangle to calculate area and perimeter

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def get_area(self):
        return self.length * self.width
    def get_perimeter(self):
        return 2 * (self.length + self.width)

rect = Rectangle(10, 5)
print(f"Rectangle Area: {rect.get_area()}, Perimeter: {rect.get_perimeter()}")
```

### 28. Create a class Employee with salary and method to increase salary

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def increase_salary(self, percentage):
        self.salary += self.salary * (percentage / 100)

emp = Employee("Alice", 50000)
emp.increase_salary(10)
print(f"Employee {emp.name} New Salary: {emp.salary}")
```

### 29. Create a class ShoppingCart using dictionary (item: price) and calculate total bill

```python
class ShoppingCart:
    def __init__(self):
        self.items = {}
    def add_item(self, item, price):
        self.items[item] = price
    def get_total(self):
        return sum(self.items.values())

cart = ShoppingCart()
cart.add_item("Book", 15)
cart.add_item("Pen", 2)
print(f"Total Bill: ${cart.get_total()}")
```

### 30. Create a class Library to store books in a list and add/remove books

```python
class Library:
    def __init__(self):
        self.books = []
    def add_book(self, book):
        self.books.append(book)
    def remove_book(self, book):
        if book in self.books:
            self.books.remove(book)
        else:
            print(f"{book} not found in library.")
    def list_books(self):
        return self.books

lib = Library()
lib.add_book("1984")
lib.add_book("To Kill a Mockingbird")
lib.remove_book("1984")
print(f"Current books in library: {lib.list_books()}")
```
